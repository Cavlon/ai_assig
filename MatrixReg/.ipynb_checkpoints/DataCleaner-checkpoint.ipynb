```python
import pandas as pd
```

```python
fdf = pd.read_csv('data/sgemm_product.csv')
fdf
```

```python
# 1-2. MWG, NWG: per-matrix 2D tiling at workgroup level: {16, 32, 64, 128} (integer)
# 3. KWG: inner dimension of 2D tiling at workgroup level: {16, 32} (integer)
# 4-5. MDIMC, NDIMC: local workgroup size: {8, 16, 32} (integer)
# 6-7. MDIMA, NDIMB: local memory shape: {8, 16, 32} (integer)
# 8. KWI: kernel loop unrolling factor: {2, 8} (integer)
# 9-10. VWM, VWN: per-matrix vector widths for loading and storing: {1, 2, 4, 8} (integer)
# 11-12. STRM, STRN: enable stride for accessing off-chip memory within a single thread: {0, 1} (categorical)
# 13-14. SA, SB: per-matrix manual caching of the 2D workgroup tile: {0, 1} (categorical)
```

```python
fdf.dtypes
```

```python
fdf.describe()
```

```python
fdf.isna().sum()
```

```python
fdf['Time'] = (fdf['Run1 (ms)'] + fdf['Run2 (ms)'] + fdf['Run3 (ms)'] + fdf['Run4 (ms)']) / 4
fdf
```

```python
fdf.drop_duplicates(inplace=True)
fdf
```

```python
drop_cols = ['Run1 (ms)', 'Run2 (ms)', 'Run3 (ms)', 'Run4 (ms)']
df = fdf.drop(drop_cols, axis=1)
df
```

```python
replace_vals = {16:-1, 32:0, 64:1, 128:2}
df['MWG'].replace(replace_vals, inplace=True)
df['NWG'].replace(replace_vals, inplace=True)
df
```

```python
replace_vals = {16:0, 32:1}
df['KWG'].replace(replace_vals, inplace=True)
df
```

```python
replace_vals = {8:-1, 16:0, 32:1}
df['MDIMC'].replace(replace_vals, inplace=True)
df['NDIMC'].replace(replace_vals, inplace=True)
df['MDIMA'].replace(replace_vals, inplace=True)
df['NDIMB'].replace(replace_vals, inplace=True)
df
```

```python
replace_vals = {2:0, 8:1}
df['KWI'].replace(replace_vals, inplace=True)
df
```

```python
replace_vals = {1:-1, 2:0, 4:1, 8:2}
df['VWM'].replace(replace_vals, inplace=True)
df['VWN'].replace(replace_vals, inplace=True)
df
```

```python
df.to_csv('data/data_cleaned.csv', index=False)
```
